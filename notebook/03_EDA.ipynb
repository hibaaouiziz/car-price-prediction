# 03 — Exploratory Data Analysis (EDA)

## Objective

The objective of this notebook is to perform an exploratory data analysis of the cleaned Car Price Prediction dataset.

EDA helps identify patterns, understand feature distributions, detect relationships between variables, and explore the factors that may influence car prices.

## Steps

1. Import the required libraries.
2. Load the cleaned dataset.
3. Perform univariate analysis.
4. Analyze numerical feature distributions.
5. Analyze categorical feature distributions.
6. Perform bivariate analysis.
7. Investigate relationships between numerical features and the target variable.
8. Analyze car prices across categorical features.
9. Perform correlation analysis.
10. Visualize relationships using pairplots and heatmaps.


## Expected Outcome

Identify important patterns and relationships in the dataset and gain insights that can guide feature engineering and model development.


```python
### 1. Import Libraries
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
### 2. Load the Cleaned Dataset
df = pd.read_csv("../data/processed/cleaned_data.csv")
df.head()
```

```python
### 3. Perform Univariate Analysis
### 3.1 Analyze Numerical Features
for col in df.select_dtypes('number').columns:
    fig, ax = plt.subplots(1,2)
    sns.histplot(x=df[col],kde=True,ax=ax[0])
    sns.boxplot(x=df[col],ax=ax[1])
    plt.show()
```

```python
### 3.2 Analyze Categorical Features
for col in df.select_dtypes('str').columns:
    sns.countplot(x=df[col])
    plt.show()
```

```python
### 4. Numerical Features Analysis
### 4.1 Correlation Heatmap
sns.heatmap(df.corr(numeric_only=True),annot=True)
```

```python
### 4.1 Pairplot
sns.pairplot(df)
```

```python
### 5. Identify Relationships with Target (Price)
sns.boxplot(x=df['Fuel Type'],y=df['Price'])
```

```python
sns.boxplot(x=df['Transmission'],y=df['Price'])
```

```python
sns.boxplot(x=df['Make'],y=df['Price'])
```

```python
sns.boxplot(x=df['Model'],y=df['Price'])
```
